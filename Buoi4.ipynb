```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import OneHotEncoder

# ==============================================================================
# PART 1: DỰ ĐOÁN PLAY (VD1) BẰNG KNN
# Mẫu dự đoán: X = (Sunny, Cool, High, True)
# ==============================================================================

# 1. Khởi tạo dữ liệu VD1
data1 = {
    'Weather': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain', 'Rain', 'Overcast', 'Sunny', 'Sunny', 'Rain'],
    'Temp': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool', 'Cool', 'Cool', 'Mild', 'Cool', 'Mild'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'Normal'],
    'Windy': [False, True, False, False, False, True, True, False, False, False],
    'Play': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes']
}

df1 = pd.DataFrame(data1)
X1 = df1[['Weather', 'Temp', 'Humidity', 'Windy']]
y1 = df1['Play']

# 2. Mã hóa dữ liệu VD1
encoder1 = OneHotEncoder(sparse_output=False)
X1_encoded = encoder1.fit_transform(X1)

# 3. Chuẩn bị mẫu kiểm tra VD1
sample1 = pd.DataFrame([['Sunny', 'Cool', 'High', True]], columns=['Weather', 'Temp', 'Humidity', 'Windy'])
sample1_encoded = encoder1.transform(sample1)

# 4. Dự đoán VD1 với k-NN
print("==========================================")
print("     VÍ DỤ 1: DỰ ĐOÁN PLAY BẰNG KNN       ")
print("==========================================")
print("Mẫu kiểm tra X = (Sunny, Cool, High, True)")
for k in [1, 3, 5]:
    knn1 = KNeighborsClassifier(n_neighbors=k)
    knn1.fit(X1_encoded, y1)
    pred1 = knn1.predict(sample1_encoded)
    print(f"-> Với K = {k}: Dự đoán Play = {pred1[0]}")


# ==============================================================================
# PART 2: DỰ ĐOÁN BUY LAPTOP (VD2) BẰNG KNN
# Mẫu dự đoán: X = (youth, medium, yes, fair)
# ==============================================================================

# 1. Khởi tạo dữ liệu VD2
data2 = {
    'Age': ['youth', 'youth', 'middle', 'senior', 'senior', 'senior', 'middle', 'youth', 'youth', 'senior', 'middle', 'youth', 'middle', 'senior', 'youth'],
    'Income': ['high', 'high', 'high', 'medium', 'low', 'low', 'low', 'medium', 'low', 'medium', 'medium', 'medium', 'high', 'medium', 'low'],
    'Student': ['no', 'no', 'no', 'no', 'yes', 'yes', 'yes', 'no', 'yes', 'yes', 'no', 'yes', 'yes', 'no', 'no'],
    'Credit': ['fair', 'excellent', 'fair', 'fair', 'fair', 'excellent', 'excellent', 'fair', 'fair', 'fair', 'excellent', 'excellent', 'fair', 'excellent', 'fair'],
    'BuyLaptop': ['no', 'no', 'yes', 'yes', 'yes', 'no', 'yes', 'yes', 'yes', 'yes', 'yes', 'yes', 'yes', 'no', 'no']
}

df2 = pd.DataFrame(data2)
X2 = df2[['Age', 'Income', 'Student', 'Credit']]
y2 = df2['BuyLaptop']

# 2. Mã hóa dữ liệu VD2
encoder2 = OneHotEncoder(sparse_output=False)
X2_encoded = encoder2.fit_transform(X2)

# 3. Chuẩn bị mẫu kiểm tra VD2
sample2 = pd.DataFrame([['youth', 'medium', 'yes', 'fair']], columns=['Age', 'Income', 'Student', 'Credit'])
sample2_encoded = encoder2.transform(sample2)

# 4. Dự đoán VD2 với k-NN
print("\n==========================================")
print("   VÍ DỤ 2: DỰ ĐOÁN BUY LAPTOP BẰNG KNN   ")
print("==========================================")
print("Mẫu kiểm tra X = (youth, medium, yes, fair)")
for k in [1, 3, 5]:
    knn2 = KNeighborsClassifier(n_neighbors=k)
    knn2.fit(X2_encoded, y2)
    pred2 = knn2.predict(sample2_encoded)
    print(f"-> Với K = {k}: Dự đoán BuyLaptop = {pred2[0]}")
```
